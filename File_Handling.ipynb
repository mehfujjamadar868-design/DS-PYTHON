```python
#wehen you wirte something the existed data get clear or the file get empty and start writting new data everytime you run the code
i = "hi ,i'm jarvic..!"
m = "hi sir at your service"
k = "what are we doing today ?"
l = "i am fine thank you"
file = open("experiment.txt","w")
file.write(i)
file.write("\n")
file.write(m)
file.write("\n")
file.write(k)
file.write("\n")
file.write(l)
file.close()
file = open("experiment.txt","r")
print(file.read())
file.close()
```

```python
#in here we're append the binary data in the file
file = open("experiment.txt","ab")
o = b"556.55"
file.write(b"\n")
file.write(o)
file.close()
file = open("experiment.txt","r")
print(file.read())
file.close()
```

```python
#whenerver we are wringing somthing (single line or multiple line) the file get empty automatically
#here we are writing multiple line in file
lines = ["Hello\n", "Welcome back sir\n", "File Handling\n"]

file = open("myfile.txt", "w")
file.writelines(lines)  # Writes multiple lines
file.close()
file = open("myfile.txt","r")
print(file.read())
file.close()
```
