```python
cube = lambda a,b,c : a * b + c
```

```python
cube(2,2,1)
```

```python
solo = lambda j,k,l,m : j-k+l/m
solo(2,2,2,2)
```

```python
# Formula: value_if_true if condition else value_if_false
mojo = lambda a, b=1: (a+1, b+1) if a != 5 else "Error: a cannot be 5"

print(mojo(5, 1))  # Output: Error: a cannot be 5
print(mojo(4, 1))  # Output: (5, 2)
print(mojo(5, 5))  # Output: (7, 2)
```

```python
numbers = [1, 2, 3, 4]
squared = list(map(lambda x: x**2, numbers))
print(squared)  # Output: [1, 4, 9, 16]
```

```python
add = lambda x, y: x + y
add(3, 7)  # Output: 10
```
